```python
import pandas as pd

df = pd.read_csv("CarPrice.csv")
df.head()
```

```python
print("\nConcise summary of the DataFrame:")
df.info()

print("\nDescriptive statistics of the numerical columns:")
print(df.describe())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
df['CarCompany'] = df['CarName'].apply(lambda x: x.split(' ')[0].lower())

# Standardize car company names
company_mapping = {
    'vw': 'volkswagen',
    'vokswagen': 'volkswagen',
    'porche': 'porsche',
    'maxda': 'mazda',
    'nissan': 'nissan',
    'toyota': 'toyota'
}

df['CarCompany'] = df['CarCompany'].replace(company_mapping)

# Map 'doornumber' to numerical values
door_mapping = {'two': 2, 'four': 4}
df['doornumber'] = df['doornumber'].replace(door_mapping)

df["doornumber"] = df["doornumber"].str.replace({"two":2,"four":4})

# Map 'cylindernumber' to numerical values
cylinder_mapping = {
    'two': 2, 'three': 3, 'four': 4, 'five': 5,
    'six': 6, 'eight': 8, 'twelve': 12
}
df['cylindernumber'] = df['cylindernumber'].replace(cylinder_mapping)

# Drop original 'CarName' and 'car_ID' columns
df = df.drop(['CarName', 'car_ID'], axis=1)

print("DataFrame after car company extraction, standardization, number mapping, and column drops:")
print(df.head())
```
